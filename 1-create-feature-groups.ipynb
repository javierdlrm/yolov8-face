# 🗂️ Step 1 · Create Feature Groups

**Feature pipeline.** Parse the WIDERFace annotations and land them in the feature store.

`data/ annotations → parse → 🪧 wider_face_files`

> ℹ️ Needs the dataset in `data/` — run `python download.py` first.

```python
import os
import create_fgs
```

### 📄 Parse annotations
Read the WIDERFace `*_bbx_gt.txt` files into train/val DataFrames (image path, bbox count, timestamps).

```python
# Resolve paths relative to this notebook's directory so it runs from any container folder
BASE_DIR = os.getcwd()

# WIDERFace annotation files + the image directories they reference
TRAIN_FILE = os.path.join(BASE_DIR, "data/wider_face_split/wider_face_train_bbx_gt.txt")
VAL_FILE   = os.path.join(BASE_DIR, "data/wider_face_split/wider_face_val_bbx_gt.txt")
TRAIN_DIR_IMAGES = os.path.join(BASE_DIR, "data/WIDER_train/images")
VAL_DIR_IMAGES   = os.path.join(BASE_DIR, "data/WIDER_val/images")

df_train = create_fgs.parse_annotation_file(TRAIN_FILE, label="train", base_dir_images=TRAIN_DIR_IMAGES)
df_val   = create_fgs.parse_annotation_file(VAL_FILE, label="val", base_dir_images=VAL_DIR_IMAGES)

print(f"Parsed {len(df_train)} train and {len(df_val)} val annotations")
df_train.head()
```

### 💾 Write to the feature store
Combine train + val and create the **`wider_face_files`** feature group.

```python
df, fg = create_fgs.build_dataset_and_write_to_hopsworks(df_train, df_val)

print(f"✅ Wrote {len(df)} rows to feature group '{fg.name}' v{fg.version}")
df.head()
```
